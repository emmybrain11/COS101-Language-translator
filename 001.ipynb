```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb

df = pd.read_excel("Attacksdata.xlsx")

fig, axes = plt.subplots(2, 1, figsize=(14, 16))

sb.countplot(data=df, y="Motive", color=sb.color_palette()[2], ax=axes[0])
axes[0].set_title("Attack Frequency by Motive", fontsize=14, fontweight='bold')

sb.countplot(data=df, y="Victim", color=sb.color_palette()[0], ax=axes[1])
axes[1].set_title("Attack Frequency by Victim", fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```
